# 02. 다기준 후보 시나리오

현재 55개 지점과 1,485개 route에서 W1 거리, W2 평균 차로 수, W3 보호구역 proxy, W4 90m 예비 경사를 normalization해 후보 graph를 만든다.

## 용어

- graph(그래프): 지점 node와 두 지점 사이 후보 route edge의 모음
- weight(가중치): W1 거리, W2 차로 수, W3 보호구역 proxy, W4 경사를 합성한 비용
- minimum spanning tree(최소신장수형도, MST): 모든 node를 가장 작은 총 비용으로 잇는 tree
- stretch(우회율): 기준 graph의 최단거리 대비 선택 구조에서의 최단거리 비율
- t-spanner(t-스패너): stretch를 일정 수준 이하로 제한하려는 sparse graph heuristic



## 현재 데이터 기준

- 기준 지점: 55개
- 무방향 route pair: 1,485개
- W1: 저장된 네이버 Directions `data/routes`, API 재호출 없음
- W2: 국가교통정보센터 표준 노드·링크 기반 평균 차로 수, 검증 PASS
- W3: 보호구역 시설점 300m buffer proxy, 최종 GEOM 기반 W3 아님
- W4: 90m 공개 DEM 기반 예비 경사, 최종 설계용 고해상도 W4 아님
